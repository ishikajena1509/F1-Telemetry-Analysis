```python
print("Sector Times")
print("-" * 40)

print("Antonelli")
print("Sector 1:", antonelli_fastest["Sector1Time"])
print("Sector 2:", antonelli_fastest["Sector2Time"])
print("Sector 3:", antonelli_fastest["Sector3Time"])

print()

print("Russell")
print("Sector 1:", russell_fastest["Sector1Time"])
print("Sector 2:", russell_fastest["Sector2Time"])
print("Sector 3:", russell_fastest["Sector3Time"])
```

```python
a_s1 = antonelli_fastest["Sector1Time"].total_seconds()
a_s2 = antonelli_fastest["Sector2Time"].total_seconds()
a_s3 = antonelli_fastest["Sector3Time"].total_seconds()

r_s1 = russell_fastest["Sector1Time"].total_seconds()
r_s2 = russell_fastest["Sector2Time"].total_seconds()
r_s3 = russell_fastest["Sector3Time"].total_seconds()
```

```python
print("Sector Time Difference (Russell - Antonelli)")
print("-" * 45)

print("Sector 1:", round(r_s1 - a_s1, 3), "seconds")
print("Sector 2:", round(r_s2 - a_s2, 3), "seconds")
print("Sector 3:", round(r_s3 - a_s3, 3), "seconds")
```

```python
from fastf1.utils import delta_time
import matplotlib.pyplot as plt
```

```python
delta, ref_tel, compare_tel = delta_time(
    antonelli_fastest,
    russell_fastest
)
```

```python
plt.figure(figsize=(15,6))

plt.plot(
    ref_tel["Distance"],
    delta,
    linewidth=2,
    label="Russell vs Antonelli"
)

plt.axhline(0, linestyle="--")

plt.title("Delta Time Comparison")
plt.xlabel("Distance (m)")
plt.ylabel("Delta Time (s)")
plt.grid()
plt.legend()

plt.show()
```
