```python
antonelli_car_data = antonelli_fastest.get_car_data()
```

```python
antonelli_car_data = antonelli_fastest.get_car_data().add_distance()
```

```python
antonelli_car_data.head()
```

```python
antonelli_car_data.columns
```

```python
antonelli_car_data.info()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,5))
plt.plot(
    antonelli_car_data["Distance"],
    antonelli_car_data["Speed"]
)

plt.title("Antonelli - Speed vs Distance")
plt.xlabel("Distance (m)")
plt.ylabel("Speed (km/h)")
plt.grid()

plt.show()
```

```python
antonelli_car_data.loc[
    antonelli_car_data["Speed"].idxmax(),
    "Distance"
]
```

```python
antonelli_car_data.loc[
    antonelli_car_data["Speed"].idxmin(),
    "Distance"
]
```

```python
antonelli_laps = session.laps.pick_driver("ANT")
```

```python
antonelli_fastest = antonelli_laps.pick_fastest()
```

```python
antonelli_car_data = antonelli_fastest.get_car_data().add_distance()
```

```python
russell_laps = session.laps.pick_driver("RUS")
```

```python
russell_fastest = russell_laps.pick_fastest()
```

```python
russell_car_data = russell_fastest.get_car_data().add_distance()
```

```python
antonelli_car_data.head()
```

```python
russell_car_data.head()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(15,6))

plt.plot(
    antonelli_car_data["Distance"],
    antonelli_car_data["Speed"],
    label="Kimi Antonelli",
    linewidth=2
)

plt.plot(
    russell_car_data["Distance"],
    russell_car_data["Speed"],
    label="George Russell",
    linewidth=2
)

plt.title("2026 Chinese GP Qualifying\nAntonelli vs Russell - Speed Comparison")
plt.xlabel("Distance (m)")
plt.ylabel("Speed (km/h)")
plt.legend()
plt.grid()

plt.show()
```

```python
print("Fastest Lap Times")
print("------------------")
print("Antonelli:", antonelli_fastest["LapTime"])
print("Russell  :", russell_fastest["LapTime"])
```

```python
print("Top Speed")
print("---------")
print("Antonelli:", antonelli_car_data["Speed"].max(), "km/h")
print("Russell  :", russell_car_data["Speed"].max(), "km/h")
```

```python
print("Minimum Speed")
print("-------------")
print("Antonelli:", antonelli_car_data["Speed"].min(), "km/h")
print("Russell  :", russell_car_data["Speed"].min(), "km/h")
```

```python
antonelli_top = antonelli_car_data.loc[
    antonelli_car_data["Speed"].idxmax()
]

print(antonelli_top["Distance"])
```

```python
russell_top = russell_car_data.loc[
    russell_car_data["Speed"].idxmax()
]

print(russell_top["Distance"])
```

```python
plt.figure(figsize=(15,6))

plt.plot(
    antonelli_car_data["Distance"],
    antonelli_car_data["Throttle"],
    label="Antonelli",
    linewidth=2
)

plt.plot(
    russell_car_data["Distance"],
    russell_car_data["Throttle"],
    label="Russell",
    linewidth=2
)

plt.title("Antonelli vs Russell - Throttle Comparison")
plt.xlabel("Distance (m)")
plt.ylabel("Throttle (%)")
plt.legend()
plt.grid()

plt.show()
```

```python
plt.figure(figsize=(15,4))

plt.plot(
    antonelli_car_data["Distance"],
    antonelli_car_data["Brake"].astype(int),
    label="Antonelli",
    linewidth=2
)

plt.plot(
    russell_car_data["Distance"],
    russell_car_data["Brake"].astype(int),
    label="Russell",
    linewidth=2
)

plt.title("Antonelli vs Russell - Brake Comparison")
plt.xlabel("Distance (m)")
plt.ylabel("Brake (0 = Off, 1 = On)")
plt.legend()
plt.grid()

plt.show()
```
