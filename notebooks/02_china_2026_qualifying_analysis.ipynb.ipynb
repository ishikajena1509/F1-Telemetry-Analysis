```python
import matplotlib.pyplot as plt

def create_dashboard(driver1_data, driver2_data,
                     driver1_name="Driver 1",
                     driver2_name="Driver 2"):
    """
    Creates a 6-panel telemetry dashboard for two drivers.
    """

    fig, ax = plt.subplots(6, 1, figsize=(18, 20), sharex=True)

    channels = [
        ("Speed", "Speed", "km/h"),
        ("Throttle", "Throttle", "%"),
        ("Brake", "Brake", ""),
        ("nGear", "Gear", ""),
        ("RPM", "RPM", ""),
        ("DRS", "DRS", "")
    ]

    for i, (column, title, ylabel) in enumerate(channels):

        if column == "Brake":
            ax[i].plot(driver1_data["Distance"],
                       driver1_data[column].astype(int),
                       label=driver1_name)

            ax[i].plot(driver2_data["Distance"],
                       driver2_data[column].astype(int),
                       label=driver2_name)
        else:
            ax[i].plot(driver1_data["Distance"],
                       driver1_data[column],
                       label=driver1_name)

            ax[i].plot(driver2_data["Distance"],
                       driver2_data[column],
                       label=driver2_name)

        ax[i].set_title(title)
        ax[i].set_ylabel(ylabel)
        ax[i].legend()

    ax[-1].set_xlabel("Distance (m)")

    plt.tight_layout()

    return fig
```

```python
fig = create_dashboard(
    antonelli_car_data,
    russell_car_data,
    "Antonelli",
    "Russell"
)

plt.show()
```
