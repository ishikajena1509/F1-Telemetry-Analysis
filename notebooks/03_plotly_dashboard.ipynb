```python
import fastf1
import pandas as pd
import plotly.graph_objects as go
```

```python
fastf1.Cache.enable_cache("../cache")
```

```python
session = fastf1.get_session(2026, "China", "Q")
session.load()
```

```python
antonelli = session.laps.pick_drivers("ANT").pick_fastest()
russell = session.laps.pick_drivers("RUS").pick_fastest()
```

```python
antonelli_car_data = antonelli.get_car_data().add_distance()

russell_car_data = russell.get_car_data().add_distance()
```

```python
antonelli_car_data.head()
```

```python
fig = go.Figure()

fig.add_trace(
    go.Scatter(
        x=antonelli_car_data["Distance"],
        y=antonelli_car_data["Speed"],
        mode="lines",
        name="Antonelli"
    )
)

fig.add_trace(
    go.Scatter(
        x=russell_car_data["Distance"],
        y=russell_car_data["Speed"],
        mode="lines",
        name="Russell"
    )
)

fig.update_layout(
    title="China GP 2026 Qualifying - Speed Comparison",
    xaxis_title="Distance (m)",
    yaxis_title="Speed (km/h)",
    hovermode="x unified",
    template="plotly_white"
)

fig.show()
```

```python
from plotly.subplots import make_subplots
```

```python
fig = make_subplots(
    rows=6,
    cols=1,
    shared_xaxes=True,
    vertical_spacing=0.03,
    subplot_titles=(
        "Speed",
        "Throttle",
        "Brake",
        "Gear",
        "RPM",
        "DRS"
    )
)
```

```python
fig.add_trace(
    go.Scatter(
        x=antonelli_car_data["Distance"],
        y=antonelli_car_data["Speed"],
        name="Antonelli",
        mode="lines"
    ),
    row=1,
    col=1
)

fig.add_trace(
    go.Scatter(
        x=russell_car_data["Distance"],
        y=russell_car_data["Speed"],
        name="Russell",
        mode="lines"
    ),
    row=1,
    col=1
)
```

```python
# Antonelli - Throttle
fig.add_trace(
    go.Scatter(
        x=antonelli_car_data["Distance"],
        y=antonelli_car_data["Throttle"],
        name="Antonelli",
        mode="lines",
        showlegend=False
    ),
    row=2,
    col=1
)

# Russell - Throttle
fig.add_trace(
    go.Scatter(
        x=russell_car_data["Distance"],
        y=russell_car_data["Throttle"],
        name="Russell",
        mode="lines",
        showlegend=False
    ),
    row=2,
    col=1
)

# Y-axis label for Throttle
fig.update_yaxes(title_text="Throttle (%)", row=2, col=1)
```

```python
# Antonelli - Brake
fig.add_trace(
    go.Scatter(
        x=antonelli_car_data["Distance"],
        y=antonelli_car_data["Brake"].astype(int),
        name="Antonelli",
        mode="lines",
        showlegend=False
    ),
    row=3,
    col=1
)

# Russell - Brake
fig.add_trace(
    go.Scatter(
        x=russell_car_data["Distance"],
        y=russell_car_data["Brake"].astype(int),
        name="Russell",
        mode="lines",
        showlegend=False
    ),
    row=3,
    col=1
)

fig.update_yaxes(title_text="Brake", row=3, col=1)
```

```python
# Antonelli - Gear
fig.add_trace(
    go.Scatter(
        x=antonelli_car_data["Distance"],
        y=antonelli_car_data["nGear"],
        name="Antonelli",
        mode="lines",
        showlegend=False
    ),
    row=4,
    col=1
)

# Russell - Gear
fig.add_trace(
    go.Scatter(
        x=russell_car_data["Distance"],
        y=russell_car_data["nGear"],
        name="Russell",
        mode="lines",
        showlegend=False
    ),
    row=4,
    col=1
)

fig.update_yaxes(title_text="Gear", row=4, col=1)
```

```python
# Antonelli - RPM
fig.add_trace(
    go.Scatter(
        x=antonelli_car_data["Distance"],
        y=antonelli_car_data["RPM"],
        name="Antonelli",
        mode="lines",
        showlegend=False
    ),
    row=5,
    col=1
)

# Russell - RPM
fig.add_trace(
    go.Scatter(
        x=russell_car_data["Distance"],
        y=russell_car_data["RPM"],
        name="Russell",
        mode="lines",
        showlegend=False
    ),
    row=5,
    col=1
)

fig.update_yaxes(title_text="RPM", row=5, col=1)
```

```python
# Antonelli - DRS
fig.add_trace(
    go.Scatter(
        x=antonelli_car_data["Distance"],
        y=antonelli_car_data["DRS"],
        name="Antonelli",
        mode="lines",
        showlegend=False
    ),
    row=6,
    col=1
)

# Russell - DRS
fig.add_trace(
    go.Scatter(
        x=russell_car_data["Distance"],
        y=russell_car_data["DRS"],
        name="Russell",
        mode="lines",
        showlegend=False
    ),
    row=6,
    col=1
)

fig.update_yaxes(title_text="DRS", row=6, col=1)
```

```python
fig.update_layout(
    title="China GP 2026 Qualifying - Interactive Telemetry Dashboard",
    height=1200,
    hovermode="x unified",
    template="plotly_white",
    legend_title="Drivers"
)

fig.update_xaxes(title_text="Distance (m)", row=6, col=1)

fig.show()
```
